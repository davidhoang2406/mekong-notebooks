# OHLCV Analysis

Reads daily OHLCV bars from the `market-analysis` MinIO bucket (Parquet) using Spark and inspects the schema and sample rows.

**Requires:** MinIO running with data produced by the `ohlcv_daily_ingest` Spark job in `mekong-jobs`.

```python
import os
from dotenv import load_dotenv
from model.spark import SparkFactory

load_dotenv()
```

```python
factory = SparkFactory('ohlcv_analysis')
spark = factory.session
```

```python
ANALYSIS_BUCKET = os.getenv('MINIO_ANALYSIS_BUCKET', 'market-analysis')

# Spark walks the partition tree natively — never list individual files
df = spark.read.parquet(f's3a://{ANALYSIS_BUCKET}/ohlcv.bar')
df.printSchema()
df.show(truncate=False)
```

```python
factory.stop()
```
